**Punto 5**

```python
pedidos_fin_semana = df[df["is_weekend_order"] == "Yes"]

pedidos_cupon = pedidos_fin_semana[
    pedidos_fin_semana["coupon_used"] == "Yes"
]

porcentaje_cupon = (
    len(pedidos_cupon) / len(pedidos_fin_semana)
) * 100

print("Pedidos realizados en fin de semana:", len(pedidos_fin_semana))
print("Pedidos de fin de semana que utilizaron cupón:", len(pedidos_cupon))
print("Porcentaje que utilizó cupón:", round(porcentaje_cupon, 2), "%")
```

**Punto 6**

```python
inactive = df[
    (df["customer_segment"] == "Inactive Customer") &
    (df["discount_pct"] > 30)
]

print("Cantidad de pedidos:", len(inactive))
print("Monto promedio pagado:", round(inactive["total_paid_usd"].mean(), 2))
```

**Punto 8**

```python
df["ahorro_pct_real"] = (
    df["discount_amount_usd"] / df["order_value_usd"]
) * 100

df[["order_value_usd", "discount_amount_usd", "discount_pct", "ahorro_pct_real"]]
```

No necesariamente coinciden siempre. La razón es que discount_pct puede haber sido calculado o registrado con un porcentaje redondeado, mientras que ahorro_pct_real se calcula directamente a partir de:

$$ \mathtt{ahorro\_pct\_real} = \frac{\mathtt{discount\_amount\_usd}}{\mathtt{order\_value\_usd}} $$

Por eso pueden existir pequeñas diferencias por redondeo.

**Punto 9**

```python
df["compra_grande"] = df["quantity_ordered"] > 2

compras_grandes = df[df["compra_grande"] == True]

porcentaje_devoluciones = (
    (compras_grandes["returned"] == "Yes").sum()
    / len(compras_grandes)
) * 100

print("Compras grandes:", len(compras_grandes))
print("Compras grandes devueltas:", (compras_grandes["returned"] == "Yes").sum())
print("Porcentaje de compras grandes devueltas:", round(porcentaje_devoluciones, 2), "%")
```
